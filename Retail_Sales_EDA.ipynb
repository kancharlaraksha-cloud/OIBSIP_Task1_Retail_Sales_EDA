```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
quarterly_sales = df.groupby(df["Date"].dt.to_period("Q"))["Total Amount"].sum()
```

```python
quarterly_sales
```

```python
plt.figure(figsize=(10, 5))

plt.plot(quarterly_sales.index.astype(str), quarterly_sales.values, marker="o")

plt.title("Quarterly Sales Trend")
plt.xlabel("Quarter")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

### Observation

Quarterly sales show that sales increased from Q1 to Q2 in 2023, followed by a decline in Q3. Sales reached their highest level in Q4, indicating strong performance during the final quarter of 2023. The first quarter of 2024 appears much lower because the dataset contains only a partial January 2024.

```python
bins = [0, 25, 35, 45, 55, 65, 100]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]

df["Age Group"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

```python
age_group_counts = df["Age Group"].value_counts().sort_index()
age_group_counts
```

```python
plt.figure(figsize=(8, 5))

age_group_counts.plot(kind="bar")

plt.title("Customer Distribution by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Observation

The age group distribution shows that customers are spread across different age ranges. The most represented age group has the highest number of customers, while the least represented group has fewer customers. This distribution can help the business understand its main customer segments and design targeted marketing strategies.

```python
gender_counts = df["Gender"].value_counts()

gender_counts
```

```python
plt.figure(figsize=(7, 5))

gender_counts.plot(kind="bar")

plt.title("Customer Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Observation

The gender distribution shows the number of male and female customers in the dataset. The two groups have a relatively similar representation, indicating that the customer base is not heavily dominated by one gender. This information can help the business plan marketing campaigns that appeal to different customer segments.

```python
top_products = df["Product Category"].value_counts().head(10)

top_products
```

```python
plt.figure(figsize=(8, 5))

top_products.plot(kind="bar")

plt.title("Top 10 Best-Selling Product Categories")
plt.xlabel("Product Category")
plt.ylabel("Number of Sales")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

### Observation

The chart shows the sales volume across the different product categories. The category with the highest number of transactions represents the most frequently purchased category, while the category with the lowest count has fewer transactions. This can help the business identify which product categories are most popular among customers.

```python
revenue_by_category = df.groupby("Product Category")["Total Amount"].sum()

revenue_by_category
```

```python
plt.figure(figsize=(8, 5))

revenue_by_category.plot(kind="bar")

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

### Observation

The revenue chart shows the total sales generated by each product category. The category generating the highest revenue contributes the most to overall sales, while the category with the lowest revenue contributes less. This helps the business identify which categories are financially most important.

```python
plt.figure(figsize=(10, 6))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

### Observation

The correlation heatmap shows the relationships between the numerical variables. Quantity and Total Amount have a positive relationship because purchasing more units generally increases the total amount. Price per Unit also has a positive relationship with Total Amount. Age shows a relatively weak relationship with the sales-related variables.

```python
revenue_by_gender = df.groupby("Gender")["Total Amount"].sum()

revenue_by_gender
```

```python
plt.figure(figsize=(7, 5))

revenue_by_gender.plot(kind="bar")

plt.title("Total Revenue by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Revenue")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
### Observation

The revenue by gender chart compares the total revenue generated by male and female customers. The difference in revenue between the two groups provides an additional insight into customer purchasing behavior and can help the business understand which customer segment contributes more to sales.
```

# Conclusion

The exploratory data analysis of the retail sales dataset provided useful insights into sales trends, customer demographics, and product performance. Monthly and quarterly analysis showed that sales varied across different periods, with stronger sales observed during some months and quarters. The customer analysis showed the distribution of customers across different age groups and genders. Revenue analysis also helped identify the product categories contributing most to overall sales.

## Actionable Recommendations

1. **Focus on high-performing product categories:** Increase inventory and promotional activities for categories that generate the highest revenue.

2. **Use customer segmentation:** Develop targeted marketing campaigns based on customer age groups and gender to improve customer engagement.

3. **Plan inventory according to sales trends:** Use monthly and quarterly sales patterns to prepare inventory and promotions during periods of higher expected demand.

4. **Monitor sales performance regularly:** Track revenue, quantity sold, and category performance over time to identify changes in customer purchasing behavior.
